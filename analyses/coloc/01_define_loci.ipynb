```python
import pandas as pd
```

```python
input_file = f"{directory}/topSNPs.txt"
out_unmerged = f"{directory}/loci_500kb_topSNPs.tsv"
```

```python
FLANK = 500_000  # +/-500 kb flank
```

```python
df = pd.read_csv(input_file, sep="\t")

# Parse chromosome and position
def parse_snp(snp_id):
    parts = snp_id.split(":")
    chrom = parts[0]
    pos = int(parts[1])
    return chrom, pos

df["chr"], df["BP"] = zip(*df["SNP"].map(parse_snp))

# Calculate flanks
df["left_500kb"]  = (df["BP"] - FLANK).clip(lower=1).astype(int)
df["right_500kb"] = (df["BP"] + FLANK).astype(int)
df["locusID"] = df.apply(lambda x: f"{x['chr']}:{x['left_500kb']}-{x['right_500kb']}", axis=1)

# Save
df[["SNP", "rsID", "chr", "left_500kb", "right_500kb", "locusID"]].to_csv(
    out_unmerged, sep="\t", index=False
)
print(f"Saved loci: {out_unmerged}")
```
